# 01 — explore the MaleCNS subgraph
Loads the cached neuPrint pull (`data/cache`), summarises it and plots the selected neurons.

```python
import warnings; warnings.filterwarnings('ignore')
import sys; sys.path.insert(0, '..')
from fruit_fly_djent import connectome
neurons, edges = connectome.load_subgraph()
print(connectome.summary(neurons, edges))
neurons.head()
```

```python
import matplotlib.pyplot as plt, numpy as np
fig, ax = plt.subplots(1, 3, figsize=(15, 4))
neurons['consensusNt'].fillna('None').value_counts().plot.bar(ax=ax[0], title='neurotransmitter')
ax[1].hist(np.log10(neurons.pre + neurons.post), 40); ax[1].set_title('log10 synapses per neuron')
W = connectome.build_weight_matrix(neurons, edges, spectral_radius=None)
ax[2].hist(np.log10(np.abs(W).sum(axis=1).A1 + 1), 40); ax[2].set_title('log10 total |in-weight|')
plt.tight_layout()
```

```python
# 3D skeletons of a few hub neurons with NAVis (needs data/skeletons)
import navis
from fruit_fly_djent import skeletons
nl = skeletons.to_navis(body_ids=neurons.bodyId[:12])
navis.plot3d(nl, backend='plotly', width=900, height=600)
```
